# 02 — Implementação própria com NumPy

**Conformidade:** Logistic Regression e DNN são implementadas manualmente. Não usar scikit-learn, TensorFlow, Keras ou PyTorch nesta secção. Apenas NumPy/Pandas/Python para os algoritmos.

```python
import sys; sys.path.append('../')
import pandas as pd
from src.models import TfidfVectorizerNumpy, LogisticRegressionNumpy, DenseNumpy, metrics
train=pd.read_csv('../data/processed/train.csv'); val=pd.read_csv('../data/processed/validation.csv'); test=pd.read_csv('../data/processed/test.csv')
vec=TfidfVectorizerNumpy(max_features=3000).fit(train.Text)
Xtr=vec.transform(train.Text); Xv=vec.transform(val.Text); Xte=vec.transform(test.Text)
ytr=train.label.values; yv=val.label.values; yte=test.label.values
print(Xtr.shape)
```

```python
lr=LogisticRegressionNumpy(lr=.05,epochs=500,l2=1e-4).fit(Xtr,ytr)
print('Logistic Regression:', metrics(yte,lr.predict(Xte)))
```

```python
dnn=DenseNumpy(Xtr.shape[1],hidden=(128,32),lr=1e-3,epochs=250,batch=16,dropout=.25,l2=1e-4,patience=25)
dnn.fit(Xtr,ytr,Xv,yv)
print('NumPy DNN:', metrics(yte,dnn.predict(Xte)))
```

## Melhorias implementadas

- ReLU nas camadas ocultas.
- Dropout durante treino.

- Early stopping com base na validação.
